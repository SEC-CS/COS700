```python
proof_steps = extract_proof_steps("set.mm")
```

```python
branches = split_mm_by_headings("set.mm")
print(branches.keys())
```

```python
from split_metamath import get_theorem_names

branch = '02_ZF_ZERMELO-FRAENKEL_SET_THEORY'
theorems = get_theorem_names(branches[branch])
print(theorems)
```

```python
filt = proof_steps["proof_label"].isin(theorems)
df = proof_steps[filt]
```

```python
df['proof_label'].value_counts()
```

```python
# Get the value counts
counts = df['proof_label'].value_counts()

# Calculate percentage with count <= 32
threshold = 24
count_le_32 = (counts <= threshold).sum()
total_labels = len(counts)
percentage = (count_le_32 / total_labels) * 100

print(f"Labels with count <= {threshold}: {count_le_32:,} out of {total_labels:,}")
print(f"Percentage: {percentage:.2f}%")
```

```python
percents = {}
threshold = 16
for branch, mm_content in branches.items():
    theorems = get_theorem_names(mm_content)
    filt = proof_steps["proof_label"].isin(theorems)
    df = proof_steps[filt]
    counts = df['proof_label'].value_counts()
    count_le_32 = (counts <= threshold).sum()
    total_labels = len(counts)
    percentage = (count_le_32 / total_labels) * 100
    percents[branch] = (int(count_le_32), float(percentage))
```

```python
percents
```

```python
import pandas as pd
import numpy as np

# Get the value counts
counts = df['proof_label'].value_counts()

# Calculate percentiles
percentiles = [1, 5, 10, 25, 50, 75, 90, 95, 99]
percentile_values = np.percentile(counts.values, percentiles)

# Create a readable table
percentile_df = pd.DataFrame({
    'Percentile': percentiles,
    'Value': percentile_values
})
print("Percentiles of proof label frequencies:")
print(percentile_df.to_string(index=False))
print()

# Additional useful statistics
print(f"Total unique labels: {len(counts)}")
print(f"Mean frequency: {counts.mean():.2f}")
print(f"Standard deviation: {counts.std():.2f}")
print(f"Min: {counts.min()}")
print(f"Max: {counts.max()}")
print()

# What percentage of labels appear only once?
pct_once = (counts == 1).sum() / len(counts) * 100
print(f"Labels appearing only once: {(counts == 1).sum():,} ({pct_once:.1f}%)")

# What percentage of labels appear 2-5 times?
pct_2_5 = ((counts >= 2) & (counts <= 5)).sum() / len(counts) * 100
print(f"Labels appearing 2-5 times: {((counts >= 2) & (counts <= 5)).sum():,} ({pct_2_5:.1f}%)")

# What percentage appear more than 10 times?
pct_gt_10 = (counts > 10).sum() / len(counts) * 100
print(f"Labels appearing > 10 times: {(counts > 10).sum():,} ({pct_gt_10:.1f}%)")
```
